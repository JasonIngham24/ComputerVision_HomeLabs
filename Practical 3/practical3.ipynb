```python
import sys
import os
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

## Video Capture and Display

```python
cap = cv.VideoCapture(0)
if not cap.isOpened():
    print("Cannot open camera")
    exit()
while True:
    # Capture frame-by-frame
    ret, frame = cap.read()
 
    # if frame is read correctly ret is True
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    # Our operations on the frame come here
    hsv_image = cv.cvtColor(frame, cv.COLOR_BGR2HSV)
    # Define lower and upper HSV boundaries for the color green
    lower_green = np.array([35,50,50])
    upper_green = np.array([85,255,255])
    # Define lower and upper HSV boundaries for the color Red
    lower_red = np.array([0,120,70])
    upper_red = np.array([10,255,255])
    # Define lower and upper HSV boundaries for the color Orange
    lower_orange = np.array([11,120,70])
    upper_orange = np.array([24,255,255])
    # Define lower and upper HSV boundaries for the color Yellow
    lower_yellow = np.array([25,120,70])
    upper_yellow = np.array([35,255,255])
    # Define lower and upper HSV boundaries for the color Purple
    lower_purple = np.array([135,100,60])
    upper_purple = np.array([165,255,255])

    # Create masks with cv.inRange
    green_mask = cv.inRange(hsv_image, lower_green, upper_green)
    red_mask = cv.inRange(hsv_image, lower_red, upper_red)
    orange_mask = cv.inRange(hsv_image, lower_orange, upper_orange)
    yellow_mask = cv.inRange(hsv_image, lower_yellow, upper_yellow)
    purple_mask = cv.inRange(hsv_image, lower_purple, upper_purple)

    # Combine red, orange, yellow, and purple masks
    combined_mask = cv.bitwise_or(red_mask, orange_mask)
    combined_mask = cv.bitwise_or(combined_mask, yellow_mask)
    combined_mask = cv.bitwise_or(combined_mask, purple_mask)

    # Extract selected colors from the original image
    result = cv.bitwise_and(frame, frame, mask=combined_mask)

    # Display the resulting frame
    cv.imshow('frame', result)
    if cv.waitKey(1) == ord('q'):
        break
 
# When everything done, release the capture
cap.release()
cv.destroyAllWindows()
```
